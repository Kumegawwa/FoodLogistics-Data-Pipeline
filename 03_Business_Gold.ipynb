```sql
-- Criando a tabela final de consumo usando SELECT, JOIN e WHERE
-- Esta tabela responde: "Qual é o faturamento e custo de frete de entregas atrasadas por estado e categoria?"
CREATE OR REPLACE TABLE workspace.default.gold_logistica_analitica AS
SELECT 
    c.Estado,
    p.Categoria,
    COUNT(v.ID_Venda) as Total_Pedidos_Atrasados,
    ROUND(SUM(p.Preco), 2) as Faturamento_Atrasado,
    ROUND(SUM(v.Custo_Frete), 2) as Custo_Total_Frete_Atraso
FROM workspace.default.silver_vendas_logistica v
JOIN workspace.default.silver_clientes c 
    ON v.ID_Cliente = c.ID_Cliente
JOIN workspace.default.silver_produtos p 
    ON v.ID_Produto = p.ID_Produto
WHERE v.Status_Entrega = 'atrasado' -- Usamos minúsculo porque padronizamos assim na camada Silver!
GROUP BY 
    c.Estado, 
    p.Categoria;
    
-- Exibindo os resultados para garantir que funcionou
SELECT * FROM workspace.default.gold_logistica_analitica ORDER BY Total_Pedidos_Atrasados DESC;
```
