```python
# 1. Lendo os arquivos CSV do Volume usando PySpark
df_clientes = spark.read.csv("/Volumes/workspace/default/bronze/clientes.csv", header=True, inferSchema=True)
df_produtos = spark.read.csv("/Volumes/workspace/default/bronze/produtos.csv", header=True, inferSchema=True)
df_vendas = spark.read.csv("/Volumes/workspace/default/bronze/vendas_logistica.csv", header=True, inferSchema=True)

# 2. Salvando os dados como Tabelas Delta gerenciadas (Camada Bronze)
df_clientes.write.format("delta").mode("overwrite").saveAsTable("workspace.default.bronze_clientes")
df_produtos.write.format("delta").mode("overwrite").saveAsTable("workspace.default.bronze_produtos")
df_vendas.write.format("delta").mode("overwrite").saveAsTable("workspace.default.bronze_vendas_logistica")

print("Ingestão concluída com sucesso! Tabelas Delta criadas na camada Bronze.")
```
